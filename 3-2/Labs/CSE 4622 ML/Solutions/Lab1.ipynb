```python
import numpy as np

r=np.random.randint(1, 200, size=30)
print(r)
```

```python
a=0
for i in r:
  a+=i

mean = a/30
print(mean)

print (np.mean(r)) #mean
```

```python
b=0
for i in r:
  b+=(i-mean)**2

variance = b/30
print(variance)

print(np.var(r)) #variance
```

```python
sd = variance**0.5
print(sd)

print(np.std(r)) #standard deviation
```

```python
Words = ["Apple", "Banana", "Orange", "Mango"]

def check(string):
  if string in Words:
      return True

  elif string not in Words:
      return False

if(check("Banana")):
  print("Present in the list")
if(check("Kiwi")):
  print("Present in the list")
```
