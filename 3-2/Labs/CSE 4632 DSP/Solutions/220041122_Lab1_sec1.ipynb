Task 1 (a)

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
A = 3
f0 = 5
phi = np.pi/4

omega0 = 2 * np.pi * f0

t = np.arange(0, 1 + 1e-4, 1e-4)

x_t = A * np.cos(omega0 * t + phi)
```

```python
# plot x(t)
plt.figure(figsize=(12, 6))
plt.plot(t, x_t)
plt.title('Continuous-Time Sinusoid: x(t) = 3 cos(2π * 5t + π/4)')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.grid(True)
plt.show()
```

Task 1 (b)

```python
print(f"Amplitude (A): {A}")
print(f"Frequency (f0): {f0} Hz")
print(f"Angular Frequency (ω0): {omega0} rad/s")
print(f"Phase (φ): {phi} radians (or {np.degrees(phi):.2f} degrees)")
```

Task 2 (a)

```python
f0 = 10

t_cont = np.arange(0, 0.5 + 1e-4, 1e-4)
x_cont = np.cos(2 * np.pi * f0 * t_cont)

plt.figure(figsize=(12, 8))
plt.plot(t_cont, x_cont, label='Original Continuous Signal x(t)', alpha=0.7, linestyle='--')
```

```python
fs1 = 100
Ts1 = 1 / fs1
n1 = np.arange(0, 1 * fs1)
t_discrete1 = n1 * Ts1

x_n1 = np.cos(2 * np.pi * f0 * t_discrete1)

plt.stem(t_discrete1, x_n1, linefmt='b-', markerfmt='bo', basefmt=' ', label=f'Sampled at fs={fs1} Hz')
```

Task 2 (b)

```python
fs2 = 15
Ts2 = 1 / fs2

n2 = np.arange(0, int(0.5 * fs2) + 1)
t_discrete2 = n2 * Ts2

# sampled signal
x_n2 = np.cos(2 * np.pi * f0 * t_discrete2)

# plot sampled points
plt.stem(
    t_discrete2,
    x_n2,
    linefmt='r-',
    markerfmt='ro',
    basefmt=' ',
    label=f'Sampled at fs={fs2} Hz (Aliasing)'
)

# alias frequency
fa = abs(f0 - fs2)

# alias sinusoid
x_alias = np.cos(2 * np.pi * fa * t_cont)

# plot alias waveform
plt.plot(
    t_cont,
    x_alias,
    'g',
    linewidth=2,
    label=f'Alias Signal ({fa} Hz)'
)

plt.title('Aliasing Demonstration')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)
plt.show()
```

Task 2 (c)

```python
nyquist_rate = 2 * f0

print(f"The frequency of the continuous signal (f_signal) is: {f0} Hz")
print(f"The minimum sampling frequency required to avoid aliasing (Nyquist rate) is: {nyquist_rate} Hz")
```

Task 4 (a)

```python
print("-- Verification for x1[n] ---")
N1 = 8
n1_verify = np.arange(0, 3 * N1)
x1_verify = np.cos((np.pi / 4) * n1_verify)
err1 = np.max(np.abs(x1_verify[:2*N1] - x1_verify[N1:3*N1]))
print(f'x1 periodicity error: {err1:.2e}')

print("\n--- Verification for x2[n] ---")
N2 = 14
n2_verify = np.arange(0, 3 * N2)
x2_verify = np.cos((3 * np.pi / 7) * n2_verify)
err2 = np.max(np.abs(x2_verify[:2*N2] - x2_verify[N2:3*N2]))
print(f'x2 periodicity error: {err2:.2e}')

print("\n--- Verification for x3[n] (Expected: Not Periodic) ---")
# For non-periodic signals, we cannot define a N, but we can demonstrate by trying a large N
# Let's just state it's not periodic as per analytical result.
print("  x3[n] is analytically determined to be NOT periodic.")

print("\n--- Verification for x4[n] ---")
N4 = 15
n4_verify = np.arange(0, 3 * N4)
x4_verify = np.cos((2 * np.pi / 5) * n4_verify) + np.cos((2 * np.pi / 3) * n4_verify)
err4 = np.max(np.abs(x4_verify[:2*N4] - x4_verify[N4:3*N4]))
print(f'x4 periodicity error: {err4:.2e}')
```

Task 6 (a)

```python
fs_ = 500
T_duration = 0.2
f0_ = 10

t_ = np.arange(0, T_duration + 1/fs_, 1/fs_)

# the three signals
x1_t = np.cos(2 * np.pi * f0_ * t_) # phi = 0
x2_t = np.cos(2 * np.pi * f0_ * t_ + np.pi/2) # phi = pi/2
x3_t = np.cos(2 * np.pi * f0_ * t_ - np.pi/3) # phi = -pi/3
```

Task 6 (b)

```python
plt.figure(figsize=(12, 6))
plt.plot(t_, x1_t, label='x1(t) = cos(2π·10 t)', linestyle='-')
plt.plot(t_, x2_t, label='x2(t) = cos(2π·10 t + π/2)', linestyle='--')
plt.plot(t_, x3_t, label='x3(t) = cos(2π·10 t - π/3)', linestyle=':')

plt.title('Phase Shifted Sinusoids')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)
plt.xlim(0, T_duration) # Limit x-axis to the specified duration
plt.show()
```

Task 6 (c)

```python
phi2 = np.pi/2
phi3 = -np.pi/3

# time delay for x2 relative to x1
td2 = -phi2 / (2 * np.pi * f0_)
print(f"Analytical time delay for x2(t) relative to x1(t): {td2:.4f} seconds")

# time delay for x3 relative to x1
td3 = -phi3 / (2 * np.pi * f0_)
print(f"Analytical time delay for x3(t) relative to x1(t): {td3:.4f} seconds")
```

```python
if td2 < 0:
    print(f"x2(t) leads x1(t) by {-td2:.4f} seconds.")
else:
    print(f"x2(t) lags x1(t) by {td2:.4f} seconds.")

if td3 < 0:
    print(f"x3(t) leads x1(t) by {-td3:.4f} seconds.")
else:
    print(f"x3(t) lags x1(t) by {td3:.4f} seconds.")
```
